```python
import json, glob

overlap_fires = []
for path in glob.glob("/tmp/fire_metadata/**/metadata.json", recursive=True):
    with open(path) as f:
        meta = json.load(f)
    lon = (meta["spatial_bounds"][0] + meta["spatial_bounds"][2]) / 2
    lat = (meta["spatial_bounds"][1] + meta["spatial_bounds"][3]) / 2
    if 35.0 <= lat <= 42.0 and -120.0 <= lon <= -114.1:
        overlap_fires.append((path, lat, lon, meta.get("state")))

print(len(overlap_fires))
for f in overlap_fires:
    print(f)
```

```python
mismatches = []
for path in glob.glob("/tmp/fire_metadata/fires/**/metadata.json", recursive=True):
    with open(path) as f:
        meta = json.load(f)
    mtbs_state = meta["event_id"][:2]
    if meta["state"] != mtbs_state:
        mismatches.append((path, meta["state"], mtbs_state))

print(len(mismatches))
```
